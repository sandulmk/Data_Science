```python
import numpy as np 
#import pandas as pd
from sklearn.datasets import load_iris 
from sklearn.model_selection import train_test_split  
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score 
iris=load_iris()
x=iris.data
y=iris.target
#print(x)
#print(y)
#print(data)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state =42)
nb_classifier = GaussianNB()
nb_classifier.fit(x_train,y_train)
y_pred = nb_classifier.predict(x_test)
sample =[[1,10,9,11]]
accuracy = accuracy_score(y_test,y_pred)
print(accuracy)
print(f"Accuracy:{accuracy*100:.2f}%")
new = nb_classifier.predict(sample)
print(iris.target_names[new])
```

```python
import numpy as np 
import pandas as pd
#from sklearn.datasets import load_iris 
from sklearn.model_selection import train_test_split  
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score 
from sklearn.preprocessing import LabelEncoder
data =pd.read_csv('food.csv')
x=data.iloc[:,:3]
y=data.iloc[:,3]
#print(x)
#print(y)
#print(data)
le = LabelEncoder()
categorical_columns=['Ingredient']
for col in categorical_columns:
    x[col] = le.fit_transform(x[col])
y=le.fit_transform(y)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state =42)
nb_classifier = GaussianNB()
nb_classifier.fit(x_train,y_train)
y_pred = nb_classifier.predict(x_test)
sample =[[1,10,9]]
accuracy = accuracy_score(y_test,y_pred)
print(accuracy)
print(f"Accuracy:{accuracy*100:.2f}%")
new = nb_classifier.predict(sample)
print(data.FoodType[new])
```

```python
import numpy as np 
import pandas as pd
#from sklearn.datasets import load_iris 
from sklearn.model_selection import train_test_split  
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score 
from sklearn.preprocessing import LabelEncoder
data =pd.read_csv('insurance.csv')
x=data.iloc[:,:6]
y=data.iloc[:,1]
#print(x)
#print(y)
#print(data)
le = LabelEncoder()
categorical_columns=['sex','smoker','region']
for col in categorical_columns:
    x[col] = le.fit_transform(x[col])
y=le.fit_transform(y)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state =42)
nb_classifier = GaussianNB()
nb_classifier.fit(x_train,y_train)
y_pred = nb_classifier.predict(x_test)
sample =[[1,10,9,11,4,2]]
accuracy = accuracy_score(y_test,y_pred)
print(accuracy)
print(f"Accuracy:{accuracy*100:.2f}%")
new = nb_classifier.predict(sample)
print(data.expenses[new])
```

```python
import numpy as np 
import pandas as pd
#from sklearn.datasets import load_iris 
from sklearn.model_selection import train_test_split  
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score 
from sklearn.preprocessing import LabelEncoder
data =pd.read_csv('cricket.csv')
x=data.iloc[:,:5]
y=data.iloc[:,4]
#print(x)
#print(y)
#print(data)
le = LabelEncoder()
categorical_columns=['Outlook','Temp','Humidity','Windy','Play Cricket']
for col in categorical_columns:
    x[col] = le.fit_transform(x[col])
y=le.fit_transform(y)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state =42)
nb_classifier = GaussianNB()
nb_classifier.fit(x_train,y_train)
y_pred = nb_classifier.predict(x_test)
sample =[[1,10,9,11,2]]
accuracy = accuracy_score(y_test,y_pred)
print(accuracy)
print(f"Accuracy:{accuracy*100:.2f}%")
new = nb_classifier.predict(sample)
print(data["Play Cricket"][new])
```
